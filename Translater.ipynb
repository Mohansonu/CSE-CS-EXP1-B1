```python
# Import necessary libraries
from transformers import MarianMTModel, MarianTokenizer
import gradio as gr
```

```python
model_name = 'Helsinki-NLP/opus-mt-en-hi'

# Load the tokenizer and model
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)
```

```python
def translate(text):
    # Tokenize the input text
    tokenized_text = tokenizer.prepare_seq2seq_batch([text], return_tensors='pt')

    # Perform the translation
    translation = model.generate(**tokenized_text)

    # Decode the translated text
    translated_text = tokenizer.decode(translation[0], skip_special_tokens=True)

    return translated_text
```

```python
with gr.Blocks(css="""
body {
    background-size: cover;
    color: blue;
    text-align: center;
    border: 5px solid rgba(255, 255, 255, 0.3);
    box-shadow: 0px 0px 20px rgba(255, 255, 255, 0.3);
}
.author {
    position: fixed;
    left: 20px;
    top: 10px;
    font-weight: bold;
    color: black;
    font-size: 20px;
    white-space: nowrap;
    z-index: 1000;
    font-family: 'Times New Roman', serif;
    display: flex;
    align-items: center;
}
.title {
    font-family: 'Comic Sans MS', cursive, sans-serif;
    font-weight: bold;
    animation: fadeIn 2s ease-in-out;
}
.textbox {
    border: 2px solid lightSkyBlue;
    padding: 5px;
    border-radius: 5px;
    background: rgba(0, 0, 0, 0.5);
    color: white;
    transition: all 0.3s ease-in-out;
}
.textbox:focus {
    border-color: cyan;
    box-shadow: 0px 0px 10px cyan;
}
@keyframes fadeIn {
    from { opacity: 0; }
    to { opacity: 1; }
}
""") as demo:
    gr.HTML("""
    <div class='author'> Made by BANOTH MOHAN</div>
    """)
    gr.Markdown("<h2 class='title'>🌍 English to Hindi Translator</h2>")
    gr.Markdown("Enter text in English and get the Hindi translation instantly.")

    with gr.Row():
        input_box = gr.Textbox(label="Enter text to translate", placeholder="Type here...", elem_classes=["textbox"])

    with gr.Row():
        output_box = gr.Textbox(label="Translated text", elem_classes=["textbox"])

    with gr.Row():
        translate_button = gr.Button("Translate", variant="primary")
        clear_button = gr.Button("Clear")

    translate_button.click(translate, inputs=input_box, outputs=output_box)
    clear_button.click(lambda: "", inputs=None, outputs=output_box)

demo.launch()
```
